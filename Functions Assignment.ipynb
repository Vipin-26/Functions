### Q3. What are the different ways to define and call a function in Python?
- Functions can be defined using `def` keyword followed by the function name, arguments and definition, or using `lambda` keyword to define one line functions. Function calling is done using positional, keyword, or default arguments.

```python
# Using def
def square(x):
    return x * x

# Using lambda
square_lambda = lambda x: x * x

# Calling
print(square(4))               # Positional
print(square_lambda(x=5))      # Keyword
```

### Q6. Explain the concept of generators in Python and how they are defined.
- Generators are a special kind of function that generate values one at a time, using the `yield` keyword instead of `return`.

```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for num in count_up_to(3):
    print(num)
```

### Q7. What are the advantages of using generators over regular functions?
- Generators are memory efficient as values are yielded one by one. And they compute values only when needed which is useful for large data sets or infinite sequences.

```python
def big_range():
    for i in range(1000000):
        yield i  # Doesn't consume memory like a full list
```

### Q9. Explain the purpose and usage of the `map()` function in Python.

### Q10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
- `map()` applies a function to each item in an iterable and returns a new iterable with the results.
- `reduce()`applies a function that returns True or False to each item, and keeps only the items for which the function returns True.
- `filter()` applies a function cumulatively to the items in a sequence, reducing it to a single value.

```python
nums = [1, 2, 3]
squared = map(lambda x: x**2, nums)
print(list(squared))
```

```python
nums = [1, 2, 3, 4]
evens = filter(lambda x: x % 2 == 0, nums)
print(list(evens))
```

```python
from functools import reduce

nums = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, nums)
print(product)
```

### Q1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

```python
def sum_of_even_numbers(numbers):
    even_sum = 0
    for num in numbers:
        if num % 2 == 0:
            even_sum += num
    return even_sum

nums = [1, 2, 3, 4, 5, 6]
result = sum_of_even_numbers(nums)
print(result)
```

### Q4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(n):
    if n < 2:
        return False  # 0 and 1 are not prime
    for i in range(2, n - 1):  # Check up to n-1
        if n % i == 0:
            return False
    return True

print(is_prime(2))    
print(is_prime(13))   
print(is_prime(100))  
print(is_prime(199))  
```

### Q5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

```python
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.count += 1
            self.a, self.b = self.b, self.a + self.b
            return self.a

        
fib = FibonacciIterator(10)

for num in fib:
    print(num, end=' ')
```

### Q6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def powers_of_two(max_exp):
    for exp in range(max_exp + 1):
        yield 2 ** exp

for value in powers_of_two(5):
    print(value)
```

### Q7. Implement a generator function that reads a file line by line and yields each line as a string.

```python
def read_file_lines(filepath):
    with open(filepath, 'r') as file:
        for line in file:
            yield line.rstrip('\n')  # Optionally strip newline characters

for line in read_file_lines('example.txt'):
    print(line)
```

### Q9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

celsius_temperatures = [0, 10, 20, 30, 40]

fahrenheit_temperatures = list(map(celsius_to_fahrenheit, celsius_temperatures))

print(fahrenheit_temperatures)
```

### Q10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
input_string = 'Hello World!'

filtered_string = ''.join(filter(lambda char: char not in 'aeiouAEIOU', input_string))

print(filtered_string)
```
